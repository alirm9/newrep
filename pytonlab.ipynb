```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 4, 0.05)
y = np.sin(x*np.pi)

fig, ax = plt.subplots(figsize=(6, 4), constrained_layout=True)
ax.plot(x, y, label='sin(x*pi)')
ax.set_xlabel('time (s)')
ax.set_ylabel('amplitude')
ax.legend()
fig.set_facecolor('lightgray')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Visualise a 4th polynomial function
x = np.linspace(0,4)
y = 0.1 * x**4 - 0.5 * x**3 + 0.2 * x**2 + 0.1 * x + 4
fig, ax = plt.subplots( figsize = (10,6)))
plt.plot (x, y, linestyle = "--" , color = "red", linewidth = 2.5, label = "4th polynomial")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graph of 4th Polynomial Function")
plt.legend()
plt.show()
```
