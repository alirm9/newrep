```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 4, 0.05)
y = np.sin(x*np.pi)

fig, ax = plt.subplots(figsize=(6, 4), constrained_layout=True)
ax.plot(x, y, label='sin(x*pi)')
ax.set_xlabel('time (s)')
ax.set_ylabel('amplitude')
ax.legend()
fig.set_facecolor('lightgray')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Visualise a 4th polynomial function
x = np.linspace(-2,5)
y = 0.1 * x**4 - 0.5 * x**3 + 0.2 * x**2 + 0.1 * x + 4
fig, ax = plt.subplots (figsize = (10,6))
plt.plot (x, y, linestyle = "-" , color = "Green", linewidth = 2.5, label = "4th polynomial")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graph of 4th Polynomial Function")
plt.legend()
plt.show()
```

```python
# working with multiple line plots
fig, ax = plt.subplots()
x = np.linspace(-2, 5)

y1 = x - 2
y2 = x**2 - 5*x
plt.plot(x, y1, linestyle='-', color='green', linewidth=2.5, label='$ y = x - 2 $')
plt.plot(x, y2, linestyle='--', color='red', linewidth=2.5, label='$ y = x^2 - 5x $')
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graph of Multiple Polynomial Functions")
plt.legend()
plt.show()
```

```python
# working with multiple subplots (axes)
fig, ax = plt.subplots(2, figsize=(10, 8))

x = np.linspace(-2, 5)

y1 = x - 2
plt.sca(ax[0])
plt.plot(x, y1, linestyle='-', color='green', linewidth=2.5, label='$ y = x - 2 $')
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graph of Multiple Polynomial Functions")
plt.legend()

y2 = -x + 2
plt.sca(ax[1])
plt.plot(x, y2, linestyle='--', color='red', linewidth=2.5, label='$ y = -x + 2 $')
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graph of Multiple Polynomial Functions")
plt.legend()
plt.show()
```

# How to solve problem Graphically:

```python
x = np.linspace(-5, 5, 200)
y = x**2 - 2*x - 5
plt.figure(figsize=(10, 5))
plt.axhline(0, color='red', linestyle='--', label='y=0')
plt.plot(x, y, linestyle='-', color='blue', linewidth=2.5, label='$ y=x^2-2x-5 $')
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graphically solving a mathematic problem:")
plt.legend()
plt.show()
```

```python
x = np.linspace(-5, 5, 200)
y = x**2 - 2*x - 5
plt.figure(figsize=(10, 5))
plt.xlim(-2, 0)
plt.ylim(-0.5, 0.5)
plt.scatter(-1.45, 0, color='green', s=100, label='Root 1')
plt.text(-1.30, 0.05, '$x=-1.45$', fontsize=10, color='green', ha='center')
plt.axhline(0, color='red', linestyle='--', label='y=0')
plt.plot(x, y, linestyle='-', color='blue', linewidth=2.5, label='$ f(x)=x^2-2x-5 $')
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graphically solving a mathematic problem:")
plt.legend()
plt.show()
```
