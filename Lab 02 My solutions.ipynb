```python
# Generate dense x array from 0 to 10
x = np.linspace(0, 10, 1000)

f = np.sqrt(3.0 * x)
g = (x - 3.12)**2

plt.figure()
plt.plot(x, f, label='f(x)')
plt.plot(x, g, label='g(x)')
plt.title("Problem 2: Overview - Both Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
